```python

import numpy as np
import matplotlib.pyplot as plt


x = df["Year"]


precip_slope, precip_intercept = np.polyfit(x, df["precip"], 1)
temp_slope, temp_intercept = np.polyfit(x, df["skt"], 1)


precip_trend = precip_slope * x + precip_intercept
temp_trend = temp_slope * x + temp_intercept


fig, ax1 = plt.subplots(figsize=(10, 6))


color1 = "blue"
color2 = "red"


ax1.plot(x, df["precip"], marker="o", color=color1, label="Precipitation")
ax1.plot(x, precip_trend, color="blue", linestyle="--", label=f"Precipitation trend (slope={precip_slope:.2f})")


ax1.set_xlabel("Year")
ax1.set_ylabel("Precipitation", color=color1)
ax1.tick_params(axis="y", labelcolor=color1)
ax1.grid(True, alpha=0.3)


ax2 = ax1.twinx()


ax2.plot(x, df["skt"], marker="o", color=color2, label="Skin Temperature")


ax2.plot(x, temp_trend, color="red", linestyle="--", label=f"Temperature trend (slope={temp_slope:.2f})")


ax2.set_ylabel("Skin Temperature", color=color2)
ax2.tick_params(axis="y", labelcolor=color2)


lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()


ax1.legend(lines1 + lines2, labels1 + labels2, loc="best")
plt.title("Precipitation and Skin Temperature with Linear Trends")
plt.tight_layout()
plt.show()
```

```python
# This gives the statistics for the second sheet of the Excel file, similar to what was done for the first sheet. It includes the slope, standard deviation, mean, maximum, and minimum values for each region during the JJAS months (June to September). The results are saved to a CSV file named "sheet2_jjas_region_statistics.csv".

import numpy as np
import pandas as pd

df2["Time"] = pd.to_datetime(df2["Time"])

regions = ["CMZ", "Reg1", "Reg2", "Reg3", "Reg4", "Reg5"]
jjas2 = df2[df2["Time"].dt.month.isin([6, 7, 8, 9])].copy()
jjas2_ts = jjas2.groupby(jjas2["Time"].dt.year)[regions].mean()

stats = {}
years = jjas2_ts.index.to_numpy(dtype=float)

for region in regions:
    values = jjas2_ts[region]
    valid = values.notna()
    slope = np.polyfit(years[valid.to_numpy()], values[valid].to_numpy(), 1)[0]

    stats[region] = {
        "slope": slope,
        "stddev": values.std(),
        "mean": values.mean(),
        "max": values.max(),
        "min": values.min(),
    }

sheet2_stats = pd.DataFrame.from_dict(stats, orient="index")
sheet2_stats.index.name = "Region"
sheet2_stats.to_csv("sheet2_jjas_region_statistics.csv")

sheet2_stats.round(3)
```

```python
import numpy as np
import pandas as pd

df3["Time"] = pd.to_datetime(df3["Time"])

regions = ["EPO", "EIO", "EAO", "EEPO", "WEIO", "EEIO"]
jjas3 = df3[df3["Time"].dt.month.isin([6, 7, 8, 9])].copy()
jjas3_ts = jjas3.groupby(jjas3["Time"].dt.year)[regions].mean()

years = jjas3_ts.index.to_numpy(dtype=float)
stats3 = {}

for region in regions:
    values = jjas3_ts[region]
    valid = values.notna()
    rainfall = values[valid].to_numpy(dtype=float)
    region_years = years[valid.to_numpy()]
    slope = np.polyfit(region_years, rainfall, 1)[0]

    stats3[region] = {
        "slope": slope,
        "stddev": values.std(),
        "mean": values.mean(),
        "max": values.max(),
        "min": values.min(),
    }

sheet3_stats = pd.DataFrame.from_dict(stats3, orient="index")
sheet3_stats.index.name = "Region"
sheet3_stats.to_csv("sheet3_jjas_region_statistics.csv")

sheet3_stats.round(3)
```
